```python
import os
from glob import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json

plt.rcParams['font.family'] = "Times New Roman"
```

```python
metrics_json_paths = glob(f'../{os.sep}logs{os.sep}finetune_20250720{os.sep}*{os.sep}*{os.sep}*{os.sep}*{os.sep}*{os.sep}*{os.sep}*{os.sep}assessment_metrics.json')
print(f'Found {len(metrics_json_paths)} metrics JSON files.')
```

```python
metrics_records = []
for metrics_json in metrics_json_paths:
    split_path_parts = metrics_json.split(os.sep)[1:]
    for i, part in enumerate(split_path_parts):
        print(f'Part {i}: {part}')
    model = split_path_parts[2]
    pretrain_method = split_path_parts[3]
    frozen_encoder = split_path_parts[5] == 'frozenencoder_true'
    n_train = int(split_path_parts[6])
    fold = split_path_parts[7].split('_')[-1]
    
    if model == 'danet': continue
    
    metrics = json.load(open(metrics_json, 'r'))
    metrics['model'] = model
    metrics['pretrain_method'] = pretrain_method
    metrics['frozen_encoder'] = frozen_encoder
    metrics['n_train'] = n_train
    metrics['fold'] = fold
    metrics_records.append(metrics)

metrics_df = pd.DataFrame(metrics_records)
metrics_df = metrics_df.loc[(metrics_df['frozen_encoder'] == False) | ((metrics_df['frozen_encoder'] == True) & (metrics_df['model'] == 'linear_probe'))]
# metrics_df.loc[metrics_df['model'] == 'linear_probe']
display(metrics_df.sort_values(by='macro_f1_score', ascending=False))
```

```python
# at each combination n_train and byol vs byol_randinit, show the models that outperformed ImageNet baseline

# --- create pivot (if you already have ft_df_pivot, skip these two lines) ---
ft_df_pivot = finetune_df.pivot_table(
    index=['model', 'n_train'],
    columns='Pre-train Method',
    values='macro_jaccard',
    aggfunc='mean'
).reset_index()

# Ensure expected columns exist (may be missing for some combos)
expected_cols = ['ImageNet', 'BYOL', 'ImageNet + BYOL']
for c in expected_cols:
    if c not in ft_df_pivot.columns:
        ft_df_pivot[c] = np.nan

# Compute improvements vs ImageNet
ft_df_pivot['impr_BYOL_vs_ImageNet'] = ft_df_pivot['BYOL'] - ft_df_pivot['ImageNet']
ft_df_pivot['impr_ImageNet_PLUS_BYOL_vs_ImageNet'] = ft_df_pivot['ImageNet + BYOL'] - ft_df_pivot['ImageNet']

# Helper to pretty-print a small dataframe of winners
def _print_winners(df, col_impr, variant_name, pct=True, top_n=None):
    winners = df.loc[df[col_impr] > 0, ['model', 'n_train', col_impr, 'ImageNet', 'BYOL', 'ImageNet + BYOL']].copy()
    if winners.empty:
        print(f"  No models where {variant_name} > ImageNet\n")
        return
    winners = winners.sort_values(by=col_impr, ascending=False)
    if pct:
        winners[col_impr] = winners[col_impr] * 100
        winners['ImageNet_pct'] = winners['ImageNet'] * 100
        # show the relevant pretrained column value as pct
        if variant_name == 'BYOL (randinit)':
            winners['BYOL_pct'] = winners['BYOL'] * 100
            cols = ['model', 'n_train', col_impr, 'ImageNet_pct', 'BYOL_pct']
        else:
            winners['ImageNet+BYOL_pct'] = winners['ImageNet + BYOL'] * 100
            cols = ['model', 'n_train', col_impr, 'ImageNet_pct', 'ImageNet+BYOL_pct']
    else:
        cols = ['model', 'n_train', col_impr, 'ImageNet', 'BYOL', 'ImageNet + BYOL']
    if top_n:
        winners = winners.head(top_n)
    print(winners[cols].to_string(index=False))
    print()

# Iterate by n_train and report models
for n in sorted(ft_df_pivot['n_train'].unique()):
    sub = ft_df_pivot[ft_df_pivot['n_train'] == n].copy()
    # drop rows with missing ImageNet baseline (can't compare)
    sub = sub[~sub['ImageNet'].isna()]
    if sub.empty:
        print(f"n_train = {n}: no ImageNet baseline measurements available\n")
        continue

    print(f"=== n_train = {n} ===")

    # BYOL (randinit) vs ImageNet
    print(f"Models where BYOL (randinit) outperformed ImageNet (improvement in percentage points):")
    _print_winners(sub, 'impr_BYOL_vs_ImageNet', 'BYOL (randinit)')

    # ImageNet + BYOL vs ImageNet
    print(f"Models where ImageNet + BYOL outperformed ImageNet (improvement in percentage points):")
    _print_winners(sub, 'impr_ImageNet_PLUS_BYOL_vs_ImageNet', 'ImageNet + BYOL')

# Optionally: create summary tables for further inspection
byol_winners_df = ft_df_pivot.loc[ft_df_pivot['impr_BYOL_vs_ImageNet'] > 0].sort_values(
    ['n_train', 'impr_BYOL_vs_ImageNet'], ascending=[True, False])
img_byol_winners_df = ft_df_pivot.loc[ft_df_pivot['impr_ImageNet_PLUS_BYOL_vs_ImageNet'] > 0].sort_values(
    ['n_train', 'impr_ImageNet_PLUS_BYOL_vs_ImageNet'], ascending=[True, False])

# Convert improvements to percentage points for easier reading
byol_winners_df = byol_winners_df.assign(impr_pct=byol_winners_df['impr_BYOL_vs_ImageNet'] * 100)
img_byol_winners_df = img_byol_winners_df.assign(impr_pct=img_byol_winners_df['impr_ImageNet_PLUS_BYOL_vs_ImageNet'] * 100)
```

```python
import matplotlib.pyplot as plt

# colorize targets
def colorize_mask(mask, cmap=cmap):
    color_mask = np.zeros((mask.shape[0], mask.shape[1], 3), dtype=np.uint8)
    for class_index, color in cmap.items():
        color_mask[mask == class_index] = color[:3]
    return color_mask

colorized_targets = np.stack([colorize_mask(mask) for mask in targets], axis=0)
```

```python
inputs_torch = torch.from_numpy(inputs_rgb).float() / 255.0
inputs_torch = (inputs_torch - mean[None, :, None, None]) / std[None, :, None, None]

inputs_torch
```

```python
weights = ['none', 'imagenet', 'byol', 'byol_randinit']

preds = {}
preds_colorized = {}
batch_size = 128
accuracies = {}

for i, weight in enumerate(weights):
    model.load_state_dict(torch.load(f'./weights/finetune_20250829/unet/{weight}/randinit_false/frozenencoder_false/500/fold_2/best_model.pth'))
    model.eval()
    
    preds[weight] = []
    accuracies[weight] = []
    for batch_idx in range(0, len(inputs_torch), batch_size):
        inputs_batch = inputs_torch[batch_idx:batch_idx+batch_size].to('cuda')
        with torch.no_grad():
            outputs = model(inputs_batch)
            preds_batch = torch.argmax(outputs, dim=1).cpu().numpy() + 1 # add 1 to match target classes
        
        preds[weight].append(preds_batch)
        
    preds[weight] = np.concatenate(preds[weight], axis=0)
    preds_colorized[weight] = np.stack([colorize_mask(mask) for mask in preds[weight]], axis=0)
    
```

```python
import pandas as pd
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from mslandcover.config import LEGEND_CLASSES

legend_classes_list = [v for k, v in sorted(LEGEND_CLASSES.items()) if k != 0]

results_df = pd.read_csv('./logs/assessment/assessment_results.csv')

cm = confusion_matrix(
    results_df['ground_truth_class_idx'],
    results_df['predicted_class_idx'],
)

# Confusion matrix whose i-th row and j-th column entry indicates the number of samples with true label being i-th class and predicted label being j-th class.


cm_df = pd.DataFrame(cm, index=legend_classes_list, columns=legend_classes_list)

# caluculate user and producer accuracy

cm_df_full = cm_df.copy()

cm_df.loc['User Accuracy'] = (cm_df_full.values.diagonal() / cm_df_full.sum(axis=0)) * 100
cm_df['Producer Accuracy'] = (cm_df_full.values.diagonal() / cm_df_full.sum(axis=1)) * 100

# overall accuracy
cm_df.loc['User Accuracy', 'Producer Accuracy'] = (np.trace(cm_df_full.values) / np.sum(cm_df_full.values)) * 100

cm_df['Producer Accuracy'] = cm_df['Producer Accuracy'].round(2)
cm_df.loc['User Accuracy'] = cm_df.loc['User Accuracy'].round(2)

# all other values rounded to 0 decimal places
cm_df.iloc[:-1, :-1] = cm_df.iloc[:-1, :-1].astype(int).astype(str)

cm_df = cm_df.astype(str)
cm_df
```
