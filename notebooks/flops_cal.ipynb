```python
import torch
from mslandcover.models import ResNetBackbone, DANet
from calflops import calculate_flops
```

```python
os_16 = ResNetBackbone()

calculate_flops(os_16, input_shape=(1, 4, 256, 256))
```

```python
danet_os16 = DANet(os_16, 8)

calculate_flops(danet_os16, (1, 4, 256, 256))
```

```python
danet_os8 = DANet(os_8, 8)

calculate_flops(danet_os8, (1, 4, 256, 256))
```
